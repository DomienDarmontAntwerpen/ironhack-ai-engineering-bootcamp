## Now, we have to clean the html code removing words

- First we remove inline JavaScript/CSS
- Then we remove html comments. This has to be done before removing regular tags since comments can contain '>' characters
- Next we can remove the remaining tags

```python
import re

def remove_scripts_styles(text):
    text = re.sub(r'<(script|style).*?>.*?</\1>', '', text, flags=re.DOTALL)
    return text

def remove_comments(text):
    text = re.sub(r'<!--.*?-->', '', text, flags=re.DOTALL)
    return text

def remove_tags(text):
    text = re.sub(r'<.*?>', '', text)
    return text

def clean_html(text):
    text = remove_scripts_styles(text)
    text = remove_comments(text)
    text = remove_tags(text)
    return text

data["text"] = data["text"].apply(clean_html)
   
```

- Remove all the special characters
    
- Remove numbers
    
- Remove all single characters
 
- Remove single characters from the start

- Substitute multiple spaces with single space

- Remove prefixed 'b'

- Convert to Lowercase

```python
# Your code
import re

def clean_text(text):
    # 1. Remove prefixed 'b' (bv. b"tekst")
    text = re.sub(r"^b\s+", "", text)

    # 2. Lowercase
    text = text.lower()

    # 3. Remove special characters (alles behalve letters en spaties)
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # 4. Remove numbers (extra safety, al deels hierboven weg)
    text = re.sub(r'\d+', ' ', text)

    # 5. Remove single characters
    text = re.sub(r'\b[a-zA-Z]\b', ' ', text)

    # 6. Remove single characters at the start
    text = re.sub(r'^\s*[a-zA-Z]\s+', ' ', text)

    # 7. Replace multiple spaces with single space
    text = re.sub(r'\s+', ' ', text)

    # 8. Strip leading/trailing spaces
    text = text.strip()

    return text

data["text"] = data["text"].apply(clean_text)
```

## Now let's work on removing stopwords
Remove the stopwords.

```python
# Your code


stop_words = set(stopwords.words("english"))

def remove_stopwords(text):
    return " ".join([word for word in text.split() if word not in stop_words])

data["text"] = data["text"].apply(remove_stopwords)


```

## Tame Your Text with Lemmatization
Break sentences into words, then use lemmatization to reduce them to their base form (e.g., "running" becomes "run"). See how this creates cleaner data for analysis!

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])

data["text"] = data["text"].apply(lemmatize_text)
```

## Bag Of Words
Let's get the 10 top words in ham and spam messages (**EXPLORATORY DATA ANALYSIS**)

```python
from sklearn.feature_extraction.text import CountVectorizer

# splits spam en ham
spam = data[data["label"] == 1]["text"]
ham = data[data["label"] == 0]["text"]

# CountVectorizer
cv = CountVectorizer()

# spam top 10
spam_matrix = cv.fit_transform(spam)
spam_counts = pd.DataFrame(spam_matrix.sum(axis=0), columns=cv.get_feature_names_out()).T
spam_top10 = spam_counts.nlargest(10, 0)

# ham top 10
ham_matrix = cv.fit_transform(ham)
ham_counts = pd.DataFrame(ham_matrix.sum(axis=0), columns=cv.get_feature_names_out()).T
ham_top10 = ham_counts.nlargest(10, 0)

print("TOP 10 SPAM:")
print(spam_top10)
print("\nTOP 10 HAM:")
print(ham_top10)
```

## Extra features

```python
money_simbol_list = "|".join(["euro","dollar","pound","€",r"\$"])
suspicious_words = "|".join(["free","cheap","sex","money","account","bank","fund","transfer","transaction","win","deposit","password"])

X_train_df = X_train.to_frame()
X_test_df = X_test.to_frame()

X_train_df['money_mark'] = X_train_df['text'].str.contains(money_simbol_list)*1
X_train_df['suspicious_words'] = X_train_df['text'].str.contains(suspicious_words)*1
X_train_df['text_len'] = X_train_df['text'].apply(lambda x: len(x))

X_test_df['money_mark'] = X_test_df['text'].str.contains(money_simbol_list)*1
X_test_df['suspicious_words'] = X_test_df['text'].str.contains(suspicious_words)*1
X_test_df['text_len'] = X_test_df['text'].apply(lambda x: len(x))

X_train_df.head()
```

## TF-IDF

- Load the vectorizer

- Vectorize all dataset

- print the shape of the vetorized dataset

```python
# Your code
# Your code
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()

X_train_bow = cv.fit_transform(X_train)
X_test_bow = cv.transform(X_test)

print("Shape train:", X_train_bow.shape)
print("Shape test:", X_test_bow.shape)
```

## And the Train a Classifier?

```python
# Your code
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

lr = LogisticRegression()
lr.fit(X_train_bow, y_train)

y_pred_lr = lr.predict(X_test_bow)

print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print(classification_report(y_test, y_pred_lr))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# initialiseren en fitten
tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Shape train:", X_train_tfidf.shape)
print("Shape test:", X_test_tfidf.shape)

nb_tfidf = LogisticRegression()
nb_tfidf.fit(X_train_tfidf, y_train)

y_pred_tfidf = nb_tfidf.predict(X_test_tfidf)

print("Accuracy TF-IDF:", accuracy_score(y_test, y_pred_tfidf))
print(classification_report(y_test, y_pred_tfidf))
```
